```python
from pathlib import Path
import sys
import traceback

# ============================================================
# DEFINIR RUTAS DEL PROYECTO
# ============================================================

try:
    # Funciona cuando ejecutas un archivo .py
    PROJECT_ROOT = Path(__file__).resolve().parents[1]
except NameError:
    # Funciona en notebooks o Python Interactive
    PROJECT_ROOT = Path.cwd()

FORECAST_APP_DIR = PROJECT_ROOT / "forecast_app"
ENV_PATH = PROJECT_ROOT / ".env"

if str(FORECAST_APP_DIR) not in sys.path:
    sys.path.insert(0, str(FORECAST_APP_DIR))

print("PROJECT_ROOT:", PROJECT_ROOT)
print("FORECAST_APP_DIR:", FORECAST_APP_DIR)
print("ENV_PATH:", ENV_PATH)
print("Existe .env:", ENV_PATH.exists())

```
