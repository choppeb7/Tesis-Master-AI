```python
def calcular_max_meses_consecutivos_sin_venta(fila):
    """
    Calcula la máxima racha de meses consecutivos sin venta.
    """

    valores = np.array(fila, dtype=float)

    max_racha = 0
    racha_actual = 0

    for valor in valores:
        if valor <= 0:
            racha_actual += 1
            max_racha = max(max_racha, racha_actual)
        else:
            racha_actual = 0

    return max_racha
```
